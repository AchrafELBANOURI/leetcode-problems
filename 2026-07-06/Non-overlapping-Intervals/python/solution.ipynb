## **Solution Code :**

```python
def eraseOverlapIntervals(intervals: list[list[int]]) -> int:
    count = 0
    intervals.sort()
    recentEnd = intervals[0][1]

    for start , end in intervals[1:] :
        if start >= recentEnd :
            recentEnd = end
        else : 
            recentEnd = min( end , recentEnd )
            count += 1
    
    return count
```

### **CASE 1 :**  
> **Input:** `nums = [[1,2],[2,3],[3,4],[1,3]]`  
> **Output:** `1`

```python
# Test CASE 1
nums = [[1,2],[2,3],[3,4],[1,3]]
output = eraseOverlapIntervals(nums)
print(output)  # Expected: 1
```

> **Input:** `nums = [[1,2],[1,2],[1,2]]` 
> **Output:** `2`
