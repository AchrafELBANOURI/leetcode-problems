## **Solution Code :**

```python
def maxProduct(nums: list[int]) -> int:
    output = nums[0] 
    minProduct , maxProduct = 1 , 1
    for item in nums :
        tmp = item * maxProduct
        maxProduct = max(tmp , item * minProduct , item)
        minProduct = min(tmp , item * minProduct , item)
        output = max( output , maxProduct)
    return output
```

### **CASE 1 :**  
> **Input:** `nums = [2, 3, -2, 4]`  
> **Output:** `6`

```python
# Test CASE 1
nums = [2, 3, -2, 4]
output = maxProduct(nums)
print(f"Input: {nums}")
print(f"Output: {output}")  # Expected: 6
```
