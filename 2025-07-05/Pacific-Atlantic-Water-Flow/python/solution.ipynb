## **Solution Code :**

```python
def pacificAtlantic( heights : list[list[int]] ) -> list[list[int]] :
    ROWS , COLS = len(heights) , len(heights[0])
    pac_set , atl_set = set() , set()
    result = []

    def dfs( r , c , visited_set , prev_height ) : 
        if( (r,c) in visited_set or r < 0 or c < 0 
           or r == ROWS or c == COLS or heights[r][c] < prev_height ) :
            return

        visited_set.add( (r,c) )
        dfs( r+1 , c , visited_set , heights[r][c])
        dfs( r-1 , c , visited_set , heights[r][c])
        dfs( r , c+1 , visited_set , heights[r][c])
        dfs( r , c-1 , visited_set , heights[r][c])
    

    for r in range(ROWS) :
        dfs( r , 0 , pac_set , heights[r][0])
        dfs( r , COLS-1 , atl_set , heights[r][COLS-1])
    
    for c in range(COLS) :
        dfs( 0 , c , pac_set , heights[c][0])
        dfs( ROWS-1 , c , atl_set , heights[c][COLS-1])
    
    for r in range(ROWS) :
        for c in range(COLS) : 
            if (r,c) in pac_set and (r,c) in atl_set : 
                result.append([r,c])
    
    return result
```

> **Output:** `[[0,4],[1,3],[1,4],[2,2],[3,0],[3,1],[4,0]]`
