## **Solution Code :**

```python
def scoreOfString( s : str ) -> int :
    l = len(s)
    score = 0
    for i in range(l-1) : score += abs( ord(s[i+1]) - ord(s[i]))
    return score 
```

### **Test Case 1:**
> **Input:** `s = "hello"`
> **Output:** `13`

```python
# Test Case 1
s = "hello"
output = scoreOfString(s)
print("Input:", s)
print("Output:", output)  # Expected: 13
```
