## **Solution Code :**

```python
def canFinish(numCourses: int, prerequisites: list[list[int]]) -> bool:
    hm = { i : [] for i in range(numCourses)}
    for course , prereq in prerequisites : 
        hm[course].append(prereq) 
    
    visited = set()

    def dfs( currentCourse : int ) -> bool : 
        if currentCourse in visited : 
            return False
        
        if hm[currentCourse] == [] : 
            return True

        visited.add(currentCourse)
        for prereq in hm[currentCourse] :
            if not dfs(prereq) : return False
        
        hm[currentCourse] = []
        visited.remove(currentCourse)

        return True
    
    for i in range(numCourses) : 
        if not dfs(i) : return False 
    
    return True
```

### **Test Case 1:**
> **Input:** `numCourses = 2, prerequisites = [[1,0]]`
> **Output:** `true`

```python
# Test Case 1
numCourses = 2
prerequisites = [[1,0]]
output = canFinish(numCourses, prerequisites)
print("Input:", numCourses, prerequisites)
print("Output:", output)  # Expected: true
```
