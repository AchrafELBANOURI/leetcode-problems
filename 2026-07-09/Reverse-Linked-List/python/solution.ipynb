## **Solution Code :**

```python
from typing import List, Optional


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    @staticmethod
    def build(values: List[int]) -> Optional["ListNode"]:
        # Edge case
        if not values:
            return None

        # First, create all nodes
        nodes = [ListNode(value) for value in values]

        # Second, connect them
        for i in range(len(nodes) - 1):
            nodes[i].next = nodes[i + 1]

        # Return the head
        return nodes[0]

    @staticmethod
    def tearDown(head: Optional["ListNode"]) -> List[int]:
        values = []
        currNode = head

        while currNode is not None:
            values.append(currNode.val)
            currNode = currNode.next

        return values

def proccess( currNode , prevNode ) : 
    if not currNode : 
        return prevNode
    else : 
        nextNode = currNode.next
        currNode.next = prevNode
        prevNode = currNode
        currNode = nextNode
        return proccess(currNode,prevNode)

def reverseList_recursive(head: Optional[ListNode]) -> Optional[ListNode]:
    return proccess(head,None)
```

### **CASE 1 :**  
> **Input:** `values = [1,2,3,4,5]`  
> **Output:** `[5,4,3,2,1]`

```python
# Test Case 1
values = [1, 2, 3, 4, 5]
head = ListNode.build(values)
newHead = reverseList_recursive(head)
print("Result :", ListNode.tearDown(newHead))
```
