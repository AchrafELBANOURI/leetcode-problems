## **Solution Code :**

```python
def coinChange(coins: list[int], amount: int) -> int:    
    arr = [amount + 1] * (amount+1)
    arr[0] = 0

    for c in coins : 
        for i in range( c , amount + 1 ) : 
            arr[i] = min( arr[i] , arr[ i - c ] + 1 )

    return arr[amount] if arr[amount] != amount + 1 else -1
```

## **Test Cases**

### **CASE 1 :**
> **Input:** `coins = [1, 2, 5]`, `amount = 11`
> **Output:** `3`

```python
# Test CASE 1
coins = [1, 2, 5]
amount = 11
output = coinChange(coins, amount)
print(output)  # Expected: 3
```
