## **Solution Code :**

> ## **First Approach: Dynamic Programming (Bottom-Up)**

Time Complexity: O(n²)
-- Space Complexity: O(n)

```python
def lengthOfLIS(nums: list[int]) -> int:
    LIS = [1] * len(nums)
    for i in range( len(nums) - 1 , -1 , -1 ) : 
        for j in range( i+1 , len(nums) ) : 
            if nums[i] < nums[j] : 
                LIS[i] = max( LIS[i] , LIS[j] + 1 )
    return max(LIS) 
```
