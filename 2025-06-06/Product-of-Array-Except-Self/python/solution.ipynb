## **Solution Code :**

```python
def productExceptSelf(nums: list[int]) -> list[int]:
    prec = 1
    result = [1] * len(nums)
    for ind in range(len(nums)):
        result[ind] = prec
        prec *= nums[ind]
    postc = 1
    for ind in range(len(nums) - 1, -1, -1):
        result[ind] *= postc
        postc *= nums[ind]
    return result
```

### **CASE 1 :**  
> **Input:** `nums = [1, 2, 3, 4]`  
> **Output:** `[24, 12, 8, 6]`

```python
# Test CASE 1
nums = [1, 2, 3, 4]
output = productExceptSelf(nums)
print(f"Input: {nums}")
print(f"Output: {output}")  # Expected: [24, 12, 8, 6]
```
