## **Solution Code :**

```python
import numpy as np
import math 

def countBits(n: int) -> list[int]:
    output = [0]*(n+1)
    for i in range(1,n+1):
        # output[i] = output[i-(2**math.floor(math.log2(i)))]+1
        output[i] = output[ i & i-1 ] + 1
    return output
```

## **Test Cases :**

### **CASE 1 :**  
> **Input:** `n = 0`  
> **Output:** `[0]`

```python
# Test CASE 1
n = 0
expected = [0]
result = countBits(n)
print(f"Test Case 1: {result == expected}, Expected: {expected}, Got: {result}")
```
