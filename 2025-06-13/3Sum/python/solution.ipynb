## **Solution Code :**

```python
def threeSum(nums: list[int]) -> list[list[int]]:
    nums.sort()
    output = []

    for i, n in enumerate(nums):
        if n > 0 : 
            break
        if i > 0 and nums[i-1] == n:
            continue
        
        l, r = i + 1, len(nums) - 1
        while l < r: 
            three_sum = n + nums[l] + nums[r]

            if three_sum > 0:
                r -= 1
            elif three_sum < 0: 
                l += 1
            else: 
                output.append([n, nums[l], nums[r]])
                l += 1
                while l < r and nums[l] == nums[l-1]:
                    l += 1

    return output      
```

### **CASE 1 :**  
> **Input:** `nums = [-1,0,1,2,-1,-4]`  
> **Output:** `[[-1,-1,2],[-1,0,1]]`

```python
# Test CASE 1
nums = [-1, 0, 1, 2, -1, -4]
output = threeSum(nums)
print(f"Input: {nums}")
print(f"Output: {output}")  # Expected: [[-1,-1,2],[-1,0,1]]
```

### **CASE 2 :**  
> **Input:** `nums = [0,1,1]`  
> **Output:** `[]`

```python
# Test CASE 2
nums = [0, 1, 1]
output = threeSum(nums)
print(f"Input: {nums}")
print(f"Output: {output}")  # Expected: []
```
