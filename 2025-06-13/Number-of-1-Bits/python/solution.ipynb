## **Solution Code :**

```python
def hammingWeight(n: int) -> int:
    return bin(n).count('1')
```

```python
# Test CASE 1
n = 11  # Binary: 00000000000000000000000000001011
expected = 3
result = hammingWeight(n)
print(f"Test Case 1: {result == expected}, Expected: {expected}, Got: {result}")
```
