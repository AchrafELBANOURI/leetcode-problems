## **Solution Code :**

```python
# Definition for a Node.
class Node:
    def __init__(self, val = 0, neighbors = None):
        self.val = val
        self.neighbors = neighbors if neighbors is not None else []

def cloneGraph(node: Node ) -> Node :
    if not node : return None

    hm = {}

    def dfs( node : Node ) -> Node : 
        if node in hm :
            return hm[node]
        
        NewNode = Node(node.val) 
        hm[node] = NewNode

        for n in node.neighbors : 
            NewNode.neighbors.append(dfs(n))
        
        return NewNode

    return dfs(node)
        
```

### **CASE 1:**
> **Input:** `adjList = [[2,4],[1,3],[2,4],[1,3]]`
> **Output:** `[[2,4],[1,3],[2,4],[1,3]]`

```python
# Test Case 1
# Create graph for test case
node1 = Node(1)
node2 = Node(2)
node3 = Node(3)
node4 = Node(4)
node1.neighbors = [node2, node4]
node2.neighbors = [node1, node3]
node3.neighbors = [node2, node4]
node4.neighbors = [node1, node3]
output = cloneGraph(node1)
print(output.val, [n.val for n in output.neighbors])  # Expected: 1, [2, 4]
```

### **CASE 2:**
> **Input:** `adjList = [[]]`

```python
# Test Case 2
# Create graph for test case
node1 = Node(1)
output = cloneGraph(node1)
print(output.val, [n.val for n in output.neighbors])  # Expected: 1, []
```
