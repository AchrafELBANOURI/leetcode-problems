## **Solution Code :**

```python
def fizzBuzz(n: int) -> list[str]:
    res = []

    for i in range(1,n+1) :
        if i % 15 == 0 : res.append("FizzBuzz")
        elif i % 5 == 0 : res.append("Buzz")
        elif i % 3 == 0 : res.append("Fizz")
        else : res.append(f"{i}")
    
    return res
```

### **CASE 1 :**  
> **Input:** `n = 4` 
> **Output:** `['1', '2', 'Fizz', '4']`

```python
# Test CASE 1
n = 4
output = fizzBuzz(4)
print(output)
```

### **CASE 2 :**  
> **Input:** `n = 0`
> **Output:** `[]`

```python
# Test CASE 2
n = 0
output = fizzBuzz(0)
print(output)
```

### **CASE 3 :**  
> **Input:** `n = 5`  
> **Output:** `['1', '2', 'Fizz', '4', 'Buzz']`

```python
# Test CASE 3
n = 5
output = fizzBuzz(5)
print(output)
```
