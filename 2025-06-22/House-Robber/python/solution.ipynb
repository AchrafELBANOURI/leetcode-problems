## **Solution Code :**

```python
def rob( nums : list[int] ) :
    rob_next = 0 
    rob_next_plus_one = 0

    for i in range( len(nums)-1 , -1 , -1 ) :
        current = max( rob_next , nums[i] + rob_next_plus_one )
        rob_next_plus_one = rob_next
        rob_next = current
    
    return rob_next
```

### Test Cases

> **Input:** `nums = []`
> **Output:** `0`

```python
# Test Case 1: Empty list
nums = []
print(rob(nums))  # Expected output: 0
```
