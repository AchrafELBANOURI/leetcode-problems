## **Solution Code :**

```python
def merge(intervals: list[list[int]]) -> list[list[int]]:
    res = []
    intervals.sort(key = lambda i : i[0])
    mergedInterval = intervals[0]

    for i in range(1 , len(intervals)) :
        if mergedInterval[1] < intervals[i][0] : 
            res.append(mergedInterval)
            mergedInterval = intervals[i]
        else : 
            mergedInterval = [ mergedInterval[0] , max( mergedInterval[1] , intervals[i][1]) ]
    
    res.append(mergedInterval)
    return res
```

### **CASE 1 :**  
> **Input:** `nums = [[1,3],[2,6],[8,10],[15,18]]` 
> **Output:** `[[1,6],[8,10],[15,18]]`

```python
# Test CASE 1
nums = [[1,3],[2,6],[8,10],[15,18]]
output = merge(nums)
print(output)  # Expected: [[1,6],[8,10],[15,18]]
```
