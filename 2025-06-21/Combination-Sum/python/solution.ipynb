## **Solution Code :**

```python
def combinationSum( candidates : list[int] , target : int ) -> list[list[int]] :
    res = []
    n = len(candidates)
    # candidates.sort()

    def findPath( path : list[int] , currInd : int , currNodeTarget : int ) :
        if currNodeTarget == 0 : 
            res.append( path.copy() )
        
        for i in range( currInd , n ) :
            if candidates[i] > currNodeTarget : 
                continue
            if i < n - 1 and min(candidates[i:]) > currNodeTarget - candidates[i] and currNodeTarget != candidates[i] :
                continue
            path.append( candidates[i] )
            findPath( path , i , currNodeTarget - candidates[i] )
            del path[-1]
    
    findPath( [] , 0 , target )
    return res
```

### **CASE 1 :**  
> **Input:** `nums = [2,3,6,7]`, `target = 7`  
> **Output:** `[[2, 2, 3], [7]]`

```python
# Test CASE 1
nums = [2,3,6,7]
target = 7
output = combinationSum(nums, target)
print(sorted([sorted(x) for x in output]))  # Expected: [[2, 2, 3], [7]]
```
