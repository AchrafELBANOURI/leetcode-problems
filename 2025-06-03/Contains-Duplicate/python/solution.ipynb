## **Solution Code :**

```python
def containsDuplicate(nums: list[int]) -> bool:
    return len(nums) != len(set(nums))
```

### **CASE 1 :**  
> **Input:** `nums = [1, 2, 3, 1]`  
> **Output:** `true`

```python
# Test CASE 1
nums = [1, 2, 3, 1]
output = containsDuplicate(nums)
print(f"Case 1 result = {output}")  # Expected: true
```

### **CASE 2 :**  
> **Input:** `nums = [1, 2, 3, 4]`  
> **Output:** `false`

```python
# Test CASE 2
nums = [1, 2, 3, 4]
output = containsDuplicate(nums)
print(f"Case 2 result = {output}")  # Expected: false
```

### **CASE 3 :**  
> **Input:** `nums = [1, 1, 1, 3, 3, 4, 3, 2, 4, 2]`  
> **Output:** `true`

```python
# Test CASE 3
nums = [1, 1, 1, 3, 3, 4, 3, 2, 4, 2]
output = containsDuplicate(nums)
print(f"Case 3 result = {output}")  # Expected: true
```
