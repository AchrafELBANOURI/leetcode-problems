## **Solution Code :**

```python
from typing import Optional , List

class ListNode:
    def __init__(self, val: int = 0, next: Optional['ListNode'] = None):
        self.val = val
        self.next = next

    @staticmethod
    def build(values: List[int]) -> Optional['ListNode']:
        # Edge Case
        if not values:
            return None

        i = len(values) - 1
        node = None
        while i >= 0:
            curr_node = ListNode(values[i], node)
            node = curr_node
            i -= 1

        return node

    @staticmethod
    def tear_down(head: Optional['ListNode']) -> List[int]:
        values = []
        curr_node = head
        while curr_node is not None:
            values.append(curr_node.val)
            curr_node = curr_node.next

        return values

def mergeTwoLists(list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        dummy = ListNode()
        tail = dummy 

        while list1 and list2 : 
            if list1.val < list2.val : 
                tail.next = list1
                list1 = list1.next
            else :
                tail.next = list2
                list2 = list2.next
            tail = tail.next

        if list1 : 
            tail.next = list1
        elif list2 :
            tail.next = list2
            
        return dummy.next
```

### **CASE 1 :**  
> **Input:** `list1 = [1, 2, 4]`, `list2 = [1, 3, 4]`  
> **Output:** `[1,1,2,3,4,4]`

```python
# Test Case 1
print("Test Case 1:")
list1 = [1, 2, 4]
list2 = [1, 3, 4]
head1 = ListNode.build(list1)
head2 = ListNode.build(list2)
new_head = mergeTwoLists(head1, head2)
print("Result :", ListNode.tear_down(new_head))
```

### **CASE 2 :**  
> **Input:** `list1 = []`, `list2 = []`  
> **Output:** `[]`

```python
# Test Case 2
print("Test Case 2:")
list1 = []
list2 = []
head1 = ListNode.build(list1)
head2 = ListNode.build(list2)
new_head = mergeTwoLists(head1, head2)
print("Result :", ListNode.tear_down(new_head))
```
