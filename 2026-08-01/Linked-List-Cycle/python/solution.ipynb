## **Solution Code :**

```python
from typing import List, Optional

class ListNode:
    def __init__(self, val: int = 0, next: Optional['ListNode'] = None):
        self.val = val
        self.next = next

    @staticmethod
    def build(values: List[int], pos: int) -> Optional['ListNode']:
        head = None
        tail = None
        node_at_pos = None

        for i, val in enumerate(values):
            curr = ListNode(val)
            if head is None:
                head = curr
            else:
                tail.next = curr
            tail = curr

            if i == pos:
                node_at_pos = curr

        # Create cycle if pos is valid
        if pos >= 0 and node_at_pos is not None:
            tail.next = node_at_pos

        return head

def hasCycle(head: Optional[ListNode] ) -> bool:
    if not head or not head.next : return False

    slow , fast = head , head.next

    while slow != fast : 
        if not fast or not fast.next : return False
        slow = slow.next
        fast = fast.next.next

    return True
```

### **CASE 1 :**  
> **Input:** `input = [3, 2, 0, -4]`, `pos = 1`  
> **Output:** `True`

```python
# Test Case 1
print("Test Case 1:")
pos = 1
input_data = [3, 2, 0, -4]
head = ListNode.build(input_data, pos)
has_cycle =hasCycle(head)
print(f"input : {input_data}, pos = {pos} : has Cycle ? {has_cycle}")
```
