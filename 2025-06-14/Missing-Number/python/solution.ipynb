## **Solution Code :**

```python
def missingNumber(nums: list[int]) -> int:
    n = len(nums)
    sum_n = (n*(n+1))//2
    sum_nums = sum(nums)
    return sum_n - sum_nums
```

## **Test Cases :**

> **Input:** `nums = [3,0,1]`  
> **Output:** `2`

```python
# Test CASE 1
nums = [3, 0, 1]
expected = 2
result = missingNumber(nums)
print(f"Test Case 1: {result == expected}, Expected: {expected}, Got: {result}")
```
