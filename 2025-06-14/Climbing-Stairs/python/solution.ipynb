## **Solution Code :**

```python
def climbStairs(n: int) -> int:
    prev_steps = [ i for i in range(n+1)]
    
    # we can start with 0 but 0:0 , 1:1 , 2:2 , 3:3 are already correct
    for i in range(4 , n+1) : # so let's begin with 4
        prev_steps[i] = prev_steps[i-1] + prev_steps[i-2]
    return prev_steps[n]
```
