## **Solution Code :**

```python
def uniquePaths( m : int , n : int ) : 
    if n == 1 and m == 1 : return 1
    if n == 0 or m == 0 : return 0

    dp = [[1 for _ in range(n)] for _ in range(m)]

    for i in range(m-2 , -1 , -1) : 
        for j in range(n-2 , -1 , -1 ) :
            dp[i][j] = dp[i+1][j] + dp[i][j+1]
        
    return dp[0][0]    
```
