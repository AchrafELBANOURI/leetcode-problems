## **Solution Code :**

```python
def findMin(nums: list[int]) -> int:
    if len(nums) == 1:
        return nums[0]

    leftMost, rightMost = 0, len(nums) - 1

    if nums[rightMost] > nums[0]:
        return nums[0]

    while leftMost < rightMost:
        medium = (leftMost + rightMost) // 2
        
        if nums[medium] > nums[medium + 1]: return nums[medium + 1]
        if nums[medium] < nums[medium - 1]: return nums[medium]
        
        if nums[medium] > nums[0]: leftMost = medium + 1
        else: rightMost = medium - 1
            
    return nums[leftMost]
```

### **CASE 1 :**  
> **Input:** `nums = [3,4,5,1,2]`  
> **Output:** `1`

```python
# Test CASE 1
nums = [3, 4, 5, 1, 2]
output = findMin(nums)
print(f"Input: {nums}")
print(f"Output: {output}")  # Expected: 1
```
