## **Solution Code :**

```python
def wordBreak(s : str , wordDict : list[str]) -> bool : 
    dp = [False] * ( len(s) + 1 )
    dp[-1] = True

    for i in range( len(s)-1 , -1 , -1 ) :
        for w in wordDict : 
            if i + len(w) <= len(s) and s[i : i+len(w)] == w : 
                dp[i] = dp[ i + len(w) ]
            if dp[i] : 
                break
    
    return dp[0]
```

### **Test Cases**

### **CASE 1 :**
> **Input:** `s = "leetcode"`, `wordDict = ["leet", "code"]`
> **Output:** `true`

```python
# Test CASE 1
s = "leetcode"
wordDict = ["leet", "code"]
output = wordBreak(s, wordDict)
print(output)  # Expected: true
```
