## **Setup & Solution Code:**

```python
from collections import Counter
from typing import List

class Solution:
    def majorityElement(self, nums: List[int]) -> List[int]:
        lst = []
        counter = Counter(nums)
        n=len(nums)

        for num , freq in counter.items() :
            if freq > n//3 :
                lst.append(num)
        
        return lst
        

def assert_test(expected: str, actual: str, test_name: str):
   if expected == actual:
       print(f"PASS: {test_name}")
   else:
       print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```

### **CASE 1:**

```python
sol = Solution()
input_data = [3,2,3]
output = sol.majorityElement(input_data)
expected = [3]
assert_test(expected, output, "Test Case 1")
```

### **CASE 2:**

```python
sol = Solution()
input_data = [1]
output = sol.majorityElement(input_data)
expected = [1]
assert_test(expected, output, "Test Case 2")
```

### **CASE 3:**

```python
sol = Solution()
input_data = [1,2]
output = sol.majorityElement(input_data)
expected = [1,2]
assert_test(expected, output, "Test Case 3")
```
