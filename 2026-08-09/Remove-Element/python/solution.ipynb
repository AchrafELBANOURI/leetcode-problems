## **Setup & Solution Code:**

```python
from typing import List

class Solution:
    def removeElement(self, nums: List[int], val: int) -> int:
        k=0

        for i in range(len(nums)):
            if nums[i] != val:
                nums[k] = nums[i]
                k+=1
        return k

def assert_test(expected: str, actual: str, test_name: str):
   if expected == actual:
       print(f"PASS: {test_name}")
   else:
       print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```

### **CASE 1:**

```python
sol = Solution()
input_data = [3,2,2,3]
input_val = 3
output = sol.removeElement(input_data,input_val);
expected = 2
assert_test(expected, output, "Test Case 1")
```

### **CASE 2:**

```python
sol = Solution()
input_data = [0,1,2,2,3,0,4,2]
input_val = 2
output = sol.removeElement(input_data,input_val)
expected = 5
assert_test(expected, output, "Test Case 2")
```
