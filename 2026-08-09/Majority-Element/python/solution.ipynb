## **Setup & Solution Code:**

```python
from collections import Counter
from typing import List

class Solution:
    def majorityElement(self, nums: List[int]) -> int :
        freq = Counter(nums)
        return max(freq , key = freq.get )
        

def assert_test(expected: str, actual: str, test_name: str):
   if expected == actual:
       print(f"PASS: {test_name}")
   else:
       print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```

### **CASE 1:**

```python
sol = Solution()
input_data = [3,2,3]
output = sol.majorityElement(input_data)
expected = 3
assert_test(expected, output, "Test Case 1")
```

### **CASE 2:**

```python
sol = Solution()
input_data = [2,2,1,1,1,2,2]
output = sol.majorityElement(input_data)
expected = 2
assert_test(expected, output, "Test Case 2")
```
