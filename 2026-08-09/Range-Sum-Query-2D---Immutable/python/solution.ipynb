## **Setup & Solution Code:**

```python
from typing import List

class NumMatrix:

    def __init__(self, matrix: List[List[int]]):
        if not matrix or not matrix[0]:
            self.prefix = []
            return

        rows, cols = len(matrix), len(matrix[0])
        # Create (rows + 1) x (cols + 1) prefix sum matrix initialized to 0
        self.prefix = [[0] * (cols + 1) for _ in range(rows + 1)]

        # Compute 2D prefix sums
        for r in range(rows):
            for c in range(cols):
                self.prefix[r + 1][c + 1] = (
                    self.prefix[r + 1][c]
                    + self.prefix[r][c + 1]
                    - self.prefix[r][c]
                    + matrix[r][c]
                )

    def sumRegion(self, row1: int, col1: int, row2: int, col2: int) -> int:
        return (
            self.prefix[row2 + 1][col2 + 1]
            - self.prefix[row1][col2 + 1]
            - self.prefix[row2 + 1][col1]
            + self.prefix[row1][col1]
        )


def assert_test(expected, actual, test_name: str):
    if expected == actual:
        print(f"PASS: {test_name}")
    else:
        print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```

### **CASE 1:**

```python
input_data = [
    [3, 0, 1, 4, 2],
    [5, 6, 3, 2, 1],
    [1, 2, 0, 1, 5],
    [4, 1, 0, 1, 7],
    [1, 0, 3, 0, 5],
]

num_matrix = NumMatrix(input_data)

# Test Case 1
input_range = [2, 1, 4, 3]
output = num_matrix.sumRegion(*input_range)
expected = 8
assert_test(expected, output, "Test Case 1")
```

### **CASE 2:**

```python
# Test Case 2
input_range = [1, 1, 2, 2]
output = num_matrix.sumRegion(*input_range)
expected = 11
assert_test(expected, output, "Test Case 2")
```

```python
# Test Case 3
input_range = [1, 2, 2, 4]
output = num_matrix.sumRegion(*input_range)
expected = 12
assert_test(expected, output, "Test Case 3")


# Immutability Test
expected_data = [
    [3, 0, 1, 4, 2],
    [5, 6, 3, 2, 1],
    [1, 2, 0, 1, 5],
    [4, 1, 0, 1, 7],
    [1, 0, 3, 0, 5],
]
assert_test(expected_data, input_data, "Immutability Test")
```
