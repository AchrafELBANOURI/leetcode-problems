## **Solution Code :**

```python
def two_sum(nums : list[int], target : int) -> list[int] :
    dict = {}
    for ind, value in enumerate(nums):
        if value in dict:
            return [dict[value], ind]
        dict[target-value] = ind
    return []
```

### **CASE 1 :**  
> **Input:** `nums = [2, 7, 11, 15]`, `target = 9`  
> **Output:** `[0, 1]`

```python
# Test CASE 1
nums = [2, 7, 11, 15]
target = 9
output = two_sum(nums, target)
print(output)  # Expected: [0, 1]
```
