## **Solution Code :**

```python
def longestConsecutive(nums):
    hs = set(nums)
    longest = 0

    for n in hs:
        if n - 1 not in hs:
            length = 1
            while n + length in hs:
                length += 1
            longest = max(longest, length)

    return longest
```

### **CASE 1 :**  

```python
# Test CASE 1
nums = [100,4,200,1,3,2]
output = longestConsecutive(nums)
print(output)  # Expected: 4
```

### **CASE 2 :**  

```python
# Test CASE 2
nums = [0,3,7,2,5,8,4,6,0,1]
output = longestConsecutive(nums)
print(output)  # Expected: 9
```
