## **Setup & Solution Code:**

```python
from typing import List, Optional


class Node:
    def __init__(self, key: int, value: int, next_node: Optional["Node"] = None):
        self.key = key
        self.value = value
        self.next = next_node


class MyHashMap:
    DEFAULT_CAPACITY = 10_007

    def __init__(self):
        self.map: List[Optional[Node]] = [None] * self.DEFAULT_CAPACITY

    def _hash(self, key: int) -> int:
        return key % len(self.map)

    def put(self, key: int, value: int) -> None:
        index_key = self._hash(key)
        current = self.map[index_key]

        while current is not None:
            if current.key == key:
                current.value = value
                return
            current = current.next

        self.map[index_key] = Node(key, value, self.map[index_key])

    def get(self, key: int) -> int:
        index_key = self._hash(key)
        current = self.map[index_key]

        while current is not None:
            if current.key == key:
                return current.value
            current = current.next

        return -1

    def remove(self, key: int) -> None:
        index_key = self._hash(key)
        current = self.map[index_key]
        prev: Optional[Node] = None

        while current is not None:
            if current.key == key:
                if prev is None:
                    self.map[index_key] = current.next
                else:
                    prev.next = current.next
                return
            prev = current
            current = current.next

class Solution : 
    def operations(ops: List[str], values: List[List[int]]) -> List[Optional[int]]:
        if ops[0] != "MyHashMap":
            raise ValueError("HashMap Operation Must be the first One")

        output: List[Optional[int]] = []
        hm = MyHashMap()
        output.append(None)

        for i in range(1, len(ops)):
            op = ops[i]
            if op == "put":
                hm.put(values[i][0], values[i][1])
                output.append(None)
            elif op == "remove":
                hm.remove(values[i][0])
                output.append(None)
            elif op == "get":
                output.append(hm.get(values[i][0]))
            else:
                raise ValueError(
                    f"Unexpected Operation Name -- Allowed Operations : HashMap - put - remove - get , index : {i}"
                )

        return output


def assert_test(expected: str, actual: str, test_name: str):
   if expected == actual:
       print(f"PASS: {test_name}")
   else:
       print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```
