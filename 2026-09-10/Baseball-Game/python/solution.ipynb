## **Setup & Solution Code:**

```python
from typing import List, Any

class Solution:
    def calPoints(self, operations: List[str]) -> int:
            scores = [0 for _ in range(len(operations))]
            currIndex = 0
        
            for op in operations:
              match op:
                case "C":
                  currIndex -= 1
                case "D":
                  scores[currIndex] = scores[currIndex - 1] * 2
                  currIndex += 1
                case "+":
                  scores[currIndex] = scores[currIndex - 1] + scores[currIndex - 2]
                  currIndex += 1
                case _:
                  scores[currIndex] = int(op)
                  currIndex += 1
        
            return int(sum(scores[:currIndex]))
        

def assert_test(expected: Any, actual: Any, test_name: str):
    if expected == actual:
        print(f"PASS: {test_name}")
    else:
        print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```

### **CASE 1:**

```python
sol = Solution()
operations = ["5","2","C","D","+"]
output = sol.calPoints(operations)
expected = 30
assert_test(expected, output, "Test Case 1")
```

### **CASE 2:**

```python
sol = Solution()
operations = ["5","-2","4","C","D","9","+","+"]
output = sol.calPoints(operations)
expected = 27
assert_test(expected, output, "Test Case 2")
```

### **CASE 3:**

```python
sol = Solution()
operations = ["1","C"]
output = sol.calPoints(operations)
expected = 0
assert_test(expected, output, "Test Case 3")
```
