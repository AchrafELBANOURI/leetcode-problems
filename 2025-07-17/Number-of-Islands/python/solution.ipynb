## **Solution Code :**

```python
def numIslands( grid :  list[list[str]]) :
    visited_set = set()
    num_isl = 0
    ROWS , COLS = len(grid) , len(grid[0])

    def dfs( r , c , visited_set ) :
        if( (r,c) in visited_set or r < 0 or c < 0
           or r == ROWS or c == COLS or grid[r][c] == '0') :
            return 
        
        visited_set.add((r,c))
        dfs(r+1 , c , visited_set)
        dfs(r-1 , c , visited_set)
        dfs(r , c+1 , visited_set)
        dfs(r , c-1 , visited_set)
    
    for r in range(ROWS) :
        for c in range(COLS) :
            if (r,c) in visited_set or grid[r][c] == '0' :
                continue
                
            num_isl += 1 
            dfs( r , c , visited_set )
    

    return num_isl
```

### **Test Case 1:**
> **Input:** `grid = [["1","1","1"],["0","1","0"],["1","1","1"]]`
> **Output:** `1`

```python
# Test Case 1
grid = [["1","1","1"],["0","1","0"],["1","1","1"]]
output = numIslands(grid)
print(output)  # Expected: 1
```
