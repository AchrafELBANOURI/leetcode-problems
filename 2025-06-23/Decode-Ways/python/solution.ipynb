## **Solution Code :**

```python
def numDecodings( s : str ) -> int :
    if not s or s[0] == '0' :
        return 0

    n = len(s)
    dp1 , dp2 = 1 , 0 

    for i in range( len(s)-1 , -1 , -1 ) :
        current = 0 
        if s[i] != '0' :
            current = dp1
            if i < n-1 and 10 <= int(s[i:i+2]) <= 26 : 
                current += dp2
        dp1 , dp2 = current , dp1
    
    return dp1
```
