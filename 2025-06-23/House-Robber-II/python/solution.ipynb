## **Solution Code :**

```python
def rob(nums: list[int]) -> int:
    robNext1 , robNextPlusOne1 = 0 , 0
    robNext2 , robNextPlusOne2 = 0 , 0

    if len(nums) == 1 : return nums[0]

    for i in range( 1 , len(nums)) :
        current1 = max( robNext1 , nums[i] + robNextPlusOne1 )
        robNextPlusOne1 = robNext1
        robNext1 = current1

        current2 = max( robNext2 , nums[len(nums)-1-i] + robNextPlusOne2 )
        robNextPlusOne2 = robNext2 
        robNext2 = current2
    
    return max( robNext1 , robNext2 )
```
