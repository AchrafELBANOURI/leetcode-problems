## **Setup & Solution Code:**

```python
from typing import List, Any

class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        profit = 0
        local_profit = 0

        for i in range(1, len(prices)):
            local_profit = prices[i] - prices[i - 1]
            if local_profit > 0:
                profit += local_profit

        return profit

def assert_test(expected: Any, actual: Any, test_name: str):
   if expected == actual:
       print(f"PASS: {test_name}")
   else:
       print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```

### **CASE 1:**

```python
sol = Solution()
prices = [7,1,5,3,6,4]
output = sol.maxProfit(prices)
expected = 7
assert_test(expected, output, "Test Case 1")
```

### **CASE 2:**

```python
sol = Solution()
prices = [1,2,3,4,5]
output = sol.maxProfit(prices)
expected = 4
assert_test(expected, output, "Test Case 2")
```

### **CASE 3:**

```python
sol = Solution()
prices = [7,6,4,3,1]
output = sol.maxProfit(prices)
expected = 0
assert_test(expected, output, "Test Case 3")
```
