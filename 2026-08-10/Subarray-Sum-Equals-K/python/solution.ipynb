## **Setup & Solution Code:**

```python
from typing import List, Any

class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        prefix_frequencies = {0: 1}
        
        prefix_sum = 0
        total = 0
        
        for num in nums:
            prefix_sum += num
        
            total += prefix_frequencies.get(prefix_sum - k, 0)
        
            prefix_frequencies[prefix_sum] = prefix_frequencies.get(prefix_sum, 0) + 1
        
        return total
        

def assert_test(expected: Any, actual: Any, test_name: str):
    if expected == actual:
        print(f"PASS: {test_name}")
    else:
        print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```

### **CASE 1:**

```python
sol = Solution()
nums = [1,1,1]
k = 2
output = sol.subarraySum(nums, k)
expected = 2
assert_test(expected, output, "Test Case 1")
```

### **CASE 2:**

```python
sol = Solution()
nums = [1,2,3]
k = 3
output = sol.subarraySum(nums, k)
expected = 2
assert_test(expected, output, "Test Case 2")
```
