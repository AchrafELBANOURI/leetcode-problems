## **Setup & Solution Code:**

```python
from typing import List , Any


class Solution:

    def firstMissingPositive(self, nums: List[int]) -> int:
        n = len(nums)

        for i in range(n):
            if nums[i] <= 0:
                nums[i] = n + 1

        for i in range(1, n + 1):
            num = abs(nums[i - 1])
            if num <= n and nums[num - 1] > 0:
                nums[num - 1] = -nums[num - 1]

        for i in range(1, n + 1):
            if nums[i - 1] >= 0:
                return i

        return n + 1

def assert_test(expected: Any, actual: Any, test_name: str):
   if expected == actual:
       print(f"PASS: {test_name}")
   else:
       print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```

### **CASE 1:**

```python
sol = Solution()
nums = [1,2,0]
output = sol.firstMissingPositive(nums)
expected = 3
assert_test(expected, output, "Test Case 1")
```

### **CASE 2:**

```python
sol = Solution()
nums = [3,4,-1,1]
output = sol.firstMissingPositive(nums)
expected = 2
assert_test(expected, output, "Test Case 2")
```

### **CASE 3:**

```python
sol = Solution()
nums = [7,8,9,11,12]
output = sol.firstMissingPositive(nums)
expected = 1
assert_test(expected, output, "Test Case 3")
```
