## **Setup & Solution Code:**

```python
from typing import List, Any
from collections import defaultdict , deque , Counter

class Solution:
        # first solution using map of deque
    def numMatchingSubseq_Deque(self, s: str, words: List[str]) -> int:
        charMap = defaultdict(deque)
        for word in words :
            charMap[word[0]].append(word)
        
        count = 0
        for c in s : 
            wordsWaiting = charMap[c]
            for _ in range(len(wordsWaiting)):
                w = wordsWaiting.popleft()
                newW = w[1:]
                if not newW:
                    count+=1
                else:
                    charMap[newW[0]].append(newW)
        
        return count

    # This One is Slightly efficient
    def numMatchingSubseq_List(self, s: str, words: List[str]) -> int:
        charMap = defaultdict(list)
        for word in words :
            charMap[word[0]].append(word)

        count = 0
        for c in s : 
            waitingWords = charMap[c]
            charMap[c] = []
            for w in waitingWords:
                newW = w[1:]
                if not newW:
                    count+=1
                else:
                    charMap[newW[0]].append(newW)

        return count

    # This is the most efficient I thought of
    def numMatchingSubseq(self, s: str, words: List[str]) -> int:
        def computeIfValid(st):
            idx = -1
            for char in st:
                idx = s.find(char, idx + 1)
                if idx == -1:
                    return 0
            return 1
        # I Used Counter instead of checking the same word multiple times
        word_counts = Counter(words)
        return sum(
            computeIfValid(word) * count for word, count in word_counts.items()
        )
    
def assert_test(expected: Any, actual: Any, test_name: str):
    if expected == actual:
        print(f"PASS: {test_name}")
    else:
        print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```

### **CASE 1:**

```python
sol = Solution()
s = "abcde"
words = ["a","bb","acd","ace"]
output = sol.numMatchingSubseq_Deque(s, words)
expected = 3
assert_test(expected, output, "Test Case 1 -- Deque")
output = sol.numMatchingSubseq_List(s,words)
assert_test(expected, output, "Test Case 1 -- List")
output = sol.numMatchingSubseq(s,words)
assert_test(expected, output, "Test Case 1 -- Counter")
```
