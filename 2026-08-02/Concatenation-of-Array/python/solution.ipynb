## **Solution Code :**

```python
from typing import List

def getConcatenation(nums: List[int]) -> List[int]:
    return nums * 2
```

### **CASE 1 :**  
> **Input:** `input = [1, 2, 1]` 
> **Output:** `[1, 2, 1, 1, 2, 1]`

```python
# Test Case 1
print("Test Case 1:")
input_data = [1, 2, 1]
output = getConcatenation(input_data)
print(f"Input : {input_data} Result : {output}")
```

### **CASE 2 :**  
> **Input:** `input = [1, 3, 2, 1]` 
> **Output:** `[1, 3, 2, 1, 1, 3, 2, 1]`

```python
# Test Case 2
print("Test Case 2:")
input_data = [1, 3, 2, 1]
output = getConcatenation(input_data)
print(f"Input : {input_data} Result : {output}")
```
