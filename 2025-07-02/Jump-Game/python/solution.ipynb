## **Solution Code :**

```python
def canJump(nums: list[int]) -> bool:
    goal = len(nums)-1

    for i in range(len(nums)-2  , -1 , -1) :
        if i + nums[i] >= goal :
            goal = i
    
    return goal == 0
```

### **CASE 1:**
> **Input:** `nums = [2,3,1,1,4]`
> **Output:** `true`

```python
# Test Case 1
nums = [2,3,1,1,4]
output = canJump(nums)
print(output)  # Expected: true
```

### **CASE 2:**
> **Input:** `nums = [3,2,1,0,4]`
> **Output:** `false`

```python
# Test Case 2
nums = [3,2,1,0,4]
output = canJump(nums)
print(output)  # Expected: False
```

### **CASE 3:**
> **Input:** `nums = [0]`
> **Output:** `true`

```python
# Test Case 3
nums = [0]
output = canJump(nums)
print(output)  # Expected: True
```
