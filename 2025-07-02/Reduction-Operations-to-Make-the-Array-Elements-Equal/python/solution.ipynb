## **Solution Code :**

```python
def reductionOperations(nums : list[int]) -> int :
    nums.sort()
    freq , result = 1 , 0
    
    for i in range(len(nums)-2 , -1 , -1):
        if nums[i] != nums[i+1]:
            result += freq
        freq+=1

    return result
```

### **CASE 1:**
> **Input:** `nums = [5,1,3]`
> **Output:** `3`

```python
# Test Case 1
nums = [5,1,3]
output = reductionOperations(nums)
print(output)  # Expected: 3
```

### **CASE 2:**
> **Input:** `nums = [1,1,1]`
> **Output:** `0`

```python
# Test Case 2
nums = [1,1,1]
output = reductionOperations(nums)
print(output)  # Expected: 0
```

### **CASE 3:**
> **Input:** `nums = [1,1,2,2,3]`
> **Output:** `4`

```python
# Test Case 3
nums = [1,1,2,2,3]
output = reductionOperations(nums)
print(output)  # Expected: 4
```
