## **Setup & Solution Code:**

```python
from typing import List , Optional

class Solution:
    class MyHashSet:
        def __init__(self):
            self.BUCKET_SIZE = 10007
            self.bucket = [[] for _ in range(self.BUCKET_SIZE)]

        def _hash(self, key: int) -> int:
            return key % self.BUCKET_SIZE

        def add(self, key: int) -> None:
            listNode = self.bucket[self._hash(key)]
            if key not in listNode :
                listNode.append(key)

        def remove(self, key: int) -> None:
            listNode = self.bucket[self._hash(key)]
            if key in listNode :
                listNode.remove(key)


        def contains(self, key: int) -> bool:
            listNode = self.bucket[self._hash(key)]
            return key in listNode
        
    @staticmethod
    def operations(
        ops: List[str], values: List[int]
    ) -> List[Optional[bool]]:
        if ops[0] != "MyHashSet" and ops[0] != "HashSet":
            raise ValueError("HashSet Operation Must be the first One")

        output: List[Optional[bool]] = []
        hs = Solution.MyHashSet()
        output.append(None)

        for i in range(1, len(ops)):
            op = ops[i]
            val = values[i]

            if op == "add":
                hs.add(val)
                output.append(None)
            elif op == "remove":
                hs.remove(val)
                output.append(None)
            elif op == "contains":
                output.append(hs.contains(val))
            else:
                raise ValueError(
                    f"Unexpected Operation Name -- Allowed Operations : HashSet - add - remove - contains , index : {i}"
                )

        return output

def assert_test(expected: str, actual: str, test_name: str):
   if expected == actual:
       print(f"PASS: {test_name}")
   else:
       print(f"FAIL: {test_name} | Expected: '{expected}', Got: '{actual}'")
```
