## **Solution Code :**

```python
def maxSubArray(nums: list[int]) -> int:
    currSum = globalSum = nums[0]
    for i in range(1,len(nums)):
        item = nums[i]
        if currSum + item > item : currSum += item 
        else : currSum = item
        if currSum > globalSum : globalSum = currSum
    return globalSum
```

### **CASE 1 :**  
> **Input:** `nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]`  
> **Output:** `6`

```python
# Test CASE 1
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
output = maxSubArray(nums)
print(f"Input: {nums}")
print(f"Output: {output}")  # Expected: 6
```

> **Output:** `1`
