## **Solution Code :**

```python
def maxProfit( prices : list[int]) -> int : 
    buy , sell = prices[0:2]
    maxProfit = 0
    for i in range(1,len(prices)-1): 
        profit = sell - buy 
        if profit >= 0 :
            if profit > maxProfit : maxProfit = profit
            sell = prices[i+1]
        else : buy , sell = prices[i:i+2]
    return maxProfit
```

### **CASE 1 :**  
> **Input:** `prices = [7, 1, 5, 3, 6, 4]`  
> **Output:** `5`

```python
# Test CASE 1
print("Test Case 1:")
prices : list[int] = [7, 1, 5, 3, 6, 4]
result0 = maxProfit(prices)
print(f"Case 1 : max Profit = {result0}")
```
